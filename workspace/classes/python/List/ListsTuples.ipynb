```python
# Write your code here
my_list = [1, 2, 3]  # Create a list
my_list.append(4)    # Add an item
my_list.remove(2)    # Remove an item
print(my_list)       # Output: [1, 3, 4]

my_tuple = (1, 2, 3)  # Create a tuple
print(my_tuple[0])    # Access first item
print(my_tuple.count(2))  # Count occurrences of 2
```
